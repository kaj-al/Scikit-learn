```python
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.impute import SimpleImputer
```

```python
data = "col_trans.xlsx"
df = pd.read_excel(data)
df.head()
```

```python
column_transformer = ColumnTransformer(
    transformers=[
        ('text', CountVectorizer(),'Review' ),
        ('social', OneHotEncoder(), ['Social Media']),
        ('num', SimpleImputer(strategy='mean'), ['age'])
    ],
    remainder='drop'
)
```

```python
transformed = column_transformer.fit_transform(df)
pd.DataFrame(transformed,columns=column_transformer.get_feature_names_out())
```

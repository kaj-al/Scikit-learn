```python
import numpy as np
import pandas as pd
from sklearn.pipeline import FeatureUnion
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
```

```python
np.random.seed(42)
data = np.random.randn(10,4)
df = pd.DataFrame(data,columns=['c1','c2','c3','y'])
df
```

```python
feature = FeatureUnion([
    ("scaler", StandardScaler()),
    ("pca", PCA(n_components=2))
])
```

```python
transformed = feature.fit_transform(df.drop(columns=['y']))
pd.DataFrame(transformed,columns=feature.get_feature_names_out())
```
